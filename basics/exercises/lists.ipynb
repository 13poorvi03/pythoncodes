```python
# Mixed list example
lst = ['poorvi', 20, 5.3, 'single', 'ghr rheti hu']

# IT companies list
it_companies = ['Facebook', 'Google', 'Microsoft', 'Apple', 'IBM', 'Oracle', 'Amazon']

# Print list and length
print(it_companies)
print(len(it_companies))

# First, middle, last items
first_item = it_companies[0]
mid_index = len(it_companies) // 2
last_item = it_companies[-1]

print(first_item)
print(it_companies[mid_index])
print(last_item)

# Append and insert
it_companies.append('Zomato')
print(it_companies)

it_companies.insert(mid_index, 'Blinkit')
print(it_companies)

# Uppercase one company
upper_companies = it_companies[1].upper()
print(upper_companies)

# Join with '#; '
joined_string = '#; '.join(it_companies)
print(joined_string)

# Check if company exists
company = "Google"
if company in it_companies:
    print(f"{company} exists in the list.")
else:
    print(f"{company} does not exist in the list.")

# Sort and reverse
it_companies.sort()
print(it_companies)

it_companies.reverse()
print(it_companies)

# Slice first 3
slice = it_companies[0:3]
print(slice)

# Slice last 3
out = it_companies[-3:]
print(out)

# Middle company/companies
n = len(it_companies)
if n % 2 == 0:
    middle = it_companies[n//2 - 1 : n//2 + 1]
else:
    middle = it_companies[n//2 : n//2 + 1]
print(middle)

# Delete first
del it_companies[0]
print(it_companies)

# Delete middle by index
del it_companies[mid_index]
print(it_companies)

# Delete middle slice (works for even/odd)
del it_companies[n//2 - 1 : n//2 + 1]
print(it_companies)

# Pop last
it_companies.pop()
print(it_companies)

# Clear all
it_companies.clear()
print(it_companies)

# Uncomment to destroy list completely
# del it_companies
# print(it_companies)   # will raise NameError
```
