```python
letter = 'p'
print(len(letter))
print(letter)

sentence = "hello world"
print(len(sentence))
print(sentence)

multiline_string = '''I am an engineering student .
i didn't find anything as rewarding as empowering people.
that's why i'm revising python again '''

print(multiline_string)

```

```python
# string concatenation
# We can connect strings together. Merging or connecting strings is called concatenation

first_name = 'poorvii'
last_name= 'sharma'
full_name = first_name +" " + last_name

print(full_name)
print(len(first_name))
print(len(last_name))
print(len(full_name))

print(len(first_name) > len(last_name)) 

# acessing index

print(first_name[0])
print(first_name[-1])

# slicing

print(first_name[0:3:1])
```

```python
# reversing string 

greeting = 'hello world '
print(greeting[::-1])

# skipping character 

lang = 'python'
print(lang[0:6:2])
```

```python
# formating 

first_name = "poorvi"
last_name = 'sharma'
lang = 'python'
formated_string = 'I am {} {}. I learn {}'.format(first_name,last_name,lang)
print(formated_string)

a=4
b=6

print('{} + {} = {}' .format(a,b,a+b))
print('{} - {} = {}'.format(a,b,a-b))
print('{} * {} = {}'.format(a,b,a*b))
print('{} / {} = {}'.format(a,b,a/b))


radius = 10
pi = 3.14
area = pi * radius ** 2
formated_string = "the area of a circle with a radius {} is {:.2f},".format(radius,area)
print(formated_string)
```

```python
# string interpolation /string formatting
a = 5
b = 3
print(f'{a} + {b} = {a+b}')
print(f'{a} - {b} = {a-b}')
```
