Numpy Indexing And Slicing Of Vector

```python
import numpy as np
```

```python
# start and end
arr = np.arange(11,21)
arr
```

```python
arr[1:5]
```

```python
arr[:]
```

```python
arr[::-2]
```

```python
arr[3::2]
```

```python
arr[4]
```

```python
arr =np.arange(1,31).reshape(6,5)
arr
```

Slicing 

```python
arr[1]
```

```python
arr[0,0]
```

```python
# slicing
arr[1:5] 
```

```python
arr[:5]
```

```python
# start,end,step
arr[0:6:2]
```

```python
arr[1::3]
```

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[:,2]
```

BOOLEAN INDEXING

```python
arr=np.arange(11,21)
arr
```

```python
bool_index=arr % 2==0
bool_index
```

```python
# only true elements 
arr =arr[bool_index] 
arr
```
