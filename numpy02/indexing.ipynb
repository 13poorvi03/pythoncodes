numpy indexing and slicing of vectors 

```python
arr = np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr[8]
```

```python
arr[2:6:2]
# slicing [start:end:step]
```

```python
arr[::-1]
```

matrix indexing

```python
arr = np.arange(1,31)
arr
```

```python
arr = arr.reshape(6,5)
arr
```

```python
arr[1]
```

```python
arr[3,2]
```

```python
arr[0:2,1:3]

# first row 0 row to 1 row the column 1st row to 2nd row 
```

```python
arr[:,2]
# printing column
```
