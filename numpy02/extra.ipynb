coverting numpy array to list 

```python
# # We can always convert an array back to a python list using tolist().

array = np.array([[1, 2, 3, 8],
                  [4, 5, 6, 7]])

nplist = array.tolist()
print(nplist)
```

```python

nums = np.array([1,2,3,4,5])
print(nums)
print("shape of nums:" , nums.shape)

numpy_two_dimensional_list = np.array([[0,1,2,4,5,6],
                                       [3,4,5,6,7,8],
                                       [6,7,8,9,10,11]])
print('shape of two dimention : ' , numpy_two_dimensional_list.shape)

print("reshape of two dimention:", numpy_two_dimensional_list.reshape(6,3))
```

```python
# data type of numpy array 

int_lists = [-3,-2,-1,0,1,2,3]
int_array = np.array(int_lists)
float_arr = np.array(int_lists,dtype = float)

print(int_array)
print(int_array.dtype)
print(float_arr)
print(float_arr.dtype)
```
