PANDAS SERIES

 a series is a one dimentional labeled array capable of holding any data type. The axis labels are collectively called the index 

CREATING SERIES 

```python
import numpy as np
import pandas as pd
```

```python
labels =['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30,40])
d ={'a':10,'b':20,'c':30}
```

```python
pd.Series(my_list)
```

```python
pd.Series(my_list,index=labels)
```

```python
pd.Series(arr)
# u cant pass 2d array in series coz series is 1 dimentional 
```

```python
pd.Series(d)
```
