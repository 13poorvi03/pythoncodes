```python
import numpy as np
import pandas as pd
```

Finding missing data

```python
data = {
    'A': [1,2,np.nan,4,5],
    'B': [6,2,3,4,5],
    'C': [1,2,3,np.nan,np.nan],
    'D': [1,np.nan,np.nan,np.nan,5]
}

# np.nan means null value 
df=pd.DataFrame(data)
df
```

```python
df.isna()
# for converting values into true and false
# for null = true and for value = false
```

```python
df.isna().sum()
# for counting null values 
```

```python
df.isna().any()
# for checking null value is present or not in each columns
```

Removing missing data
<!-- this is working on bases of rows -->

```python
df
```

```python
df.dropna()
# to removing all the null values column wise 
# when u apply this whereever null values is presentt in rows so it gonna remove the whole row 
```

```python
df
```

```python
df.dropna(thresh=3)
# u have atleast 3 non null values row wise
```

Filling the missing data
<!-- in this we use coulmn wise approch and even in selection we use columns  -->

```python
df
```

```python
df.fillna(0)
# for filling null valuee = 0
```

```python
values = {'A':0,'B':100,'C':300,'D':400}
df.fillna(value=values)

# this func is use for applying specific values on null values in columns
```

```python
df
```

```python
df.fillna(df.mean())
```
