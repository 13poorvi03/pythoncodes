```python
import numpy as np 
import pandas as pd
```

Merging 2 dataframes

```python
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    'name': ['John','Anna','Peter','Leo','Bob'],
    'department': ['HR','IT','Finance','IT','HR']
})


salaries = pd.DataFrame({
    'employee_id': [1,2,3,4,6],
    'salary': [60000,80000,65000,70000,90000],
    'bonus': [5000,10000,7000,8500,12000]
})

```

```python
employees
```

```python
salaries
```

```python
# for merging u always have that one base coulmn like employee_id w2hich is common in both

pd.merge(employees,salaries, on='employee_id',how='inner')

# We use how='inner' to keep only those rows where the employee_id exists in both DataFrames — basically the common intersection.
# Inner join returns only matching records from both tables based on the key column.
```

```python
# Outer join ka matlab hai dono DataFrames ke sabhi rows ko include karna, chahe match ho ya na ho. Agar kisi ek DataFrame me value missing ho, to Pandas us jagah NaN fill kar deta hai.

pd.merge(employees,salaries, on='employee_id',how='outer')
```

```python
pd.merge(employees,salaries, on='employee_id',how='left')
# it shows employee_id on the bases employees
```

```python
pd.merge(employees,salaries, on='employee_id',how='right')
# it shows employee_id on the bases salaries
```
